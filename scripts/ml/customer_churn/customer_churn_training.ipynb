```python
import pandas as pd

# Path ke file Parquet hasil ekspor DAG (sesuaikan dengan posisi notebook Anda)
parquet_path = "../../../datasets/parquet/ml_churn_training_dataset_latest.parquet"

# 1. Memuat data dengan Pandas
df = pd.read_parquet(parquet_path)

# 2. Cek dimensi (baris dan kolom) serta tipe data kolom
print(f"📊 Dimensi Data (Baris, Kolom): {df.shape}")
print("\n📝 Tipe Data Kolom:")
print(df.dtypes)

# 3. Menampilkan beberapa baris pertama
print("\n👀 Sampel Data:")
display(df.head(10))

# 4. Cek statistik deskriptif dasar untuk kolom numerik
print("\n📈 Statistik Deskriptif:")
display(df.describe())

# 5. Cek distribusi variabel target (misal: kolom is_churned / status churn)
if "is_churned" in df.columns:
    print("\n🎯 Distribusi Variabel Target (Churn):")
    target_dist = df["is_churned"].value_counts().sort_index()
    print(target_dist)
else:
    print("\n⚠️ Kolom target tidak ditemukan, cek nama kolom pada dtypes di atas.")
```

```python
df.isnull().sum()
```

```python
cat_cols = df.select_dtypes(include="object")
num_cols = df.select_dtypes(include="number")
num_cols.drop(columns=["user_id","is_churned"], inplace=True)
print(len(cat_cols.columns)+len(num_cols.columns))
```

```python
summary_avg = df.groupby("is_churned")[num_cols.columns].mean()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
fig, ax = plt.subplots(4,5,figsize=(12,8))
ax = ax.flatten()

for i, col in enumerate(num_cols) :
    sns.boxplot(data=df, y=col, hue="is_churned", ax=ax[i])
    
plt.tight_layout()
```

```python
fig, ax = plt.subplots(2,3,figsize=(12,8))
ax = ax.flatten()

for i, col in enumerate(cat_cols) :
    sns.countplot(data=df, x=col, hue="is_churned", ax=ax[i])
    
plt.tight_layout()
```
